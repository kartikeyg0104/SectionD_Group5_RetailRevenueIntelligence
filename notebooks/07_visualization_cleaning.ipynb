# Data Visualization - Cleaning Phase
Visualizing the data after the cleaning phase to verify data health.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned data
df = pd.read_csv('../data/interim/cleaned.csv')
df.head()
```

```python
# Visualizing missing values to confirm cleaning
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap (Post-Cleaning)')
plt.show()
```

```python
# Visualize the distributions of numerical columns
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(df['price_per_unit'], bins=30, kde=True, ax=axes[0])
axes[0].set_title('Distribution of Price Per Unit')

sns.histplot(df['quantity'], bins=20, kde=True, ax=axes[1])
axes[1].set_title('Distribution of Quantity')

sns.histplot(df['total_spent'], bins=30, kde=True, ax=axes[2])
axes[2].set_title('Distribution of Total Spent')

plt.tight_layout()
plt.show()
```

```python
# Visualize the distribution of categorical columns
categorical_cols = ['category', 'payment_method', 'location']
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, col in enumerate(categorical_cols):
    sns.countplot(data=df, x=col, ax=axes[i], order=df[col].value_counts().index)
    axes[i].set_title(f'Count of {col.title()}')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Visualizing outliers in numerical columns using boxplots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(y=df['price_per_unit'], ax=axes[0])
axes[0].set_title('Boxplot of Price Per Unit')

sns.boxplot(y=df['quantity'], ax=axes[1])
axes[1].set_title('Boxplot of Quantity')

sns.boxplot(y=df['total_spent'], ax=axes[2])
axes[2].set_title('Boxplot of Total Spent')

plt.tight_layout()
plt.show()
```
